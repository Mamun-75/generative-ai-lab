### Now we will discuss Streaming and Batch

### Streaming

Most models can stream their output content while it is being generated . by diaplaying output progressively,, streaming significantly improves user experience, particularly for longer responses . Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time

```python
model.invoke("Write me a 200 word essay on  AI")
```

it generated the essay and then it showed entire output once generation was done 

```python
for chunk in model.stream("Write me a 200 word essay on  AI"):
    print(chunk.text, end = "|", flush  = True)
```

```python
for chunk in model.stream("Why do Parrots have colorful feathers?"):
    print(chunk.text, end = "|", flush  = True)
```

===== Streaming (model.stream) নোট =====

invoke বনাম stream
- model.invoke(...) : পুরো উত্তর তৈরি হওয়া পর্যন্ত অপেক্ষা করে, তারপর একবারে দেয়
- model.stream(...) : উত্তর তৈরি হওয়ার সাথে সাথে ছোট ছোট টুকরো করে দেয়
- ChatGPT-তে লেখা যেভাবে একটু একটু করে ফুটে ওঠে, সেটাই streaming

Chunk কী
- মডেলের পাঠানো একেকটা ছোট টুকরো (একটা শব্দ, শব্দের অংশ বা চিহ্ন)
- stream() লুপে প্রতিবারে একটা নতুন chunk দেয়
- chunk.text বা chunk.content দিয়ে টুকরোর লেখাটুকু পাওয়া যায়
- কিছু chunk-এর টেক্সট খালি থাকতে পারে (reasoning মডেলে শুরুর দিকে এমন দেখা গেছে)

কোড
for chunk in model.stream("Why do Parrots have colorful feathers?"):
    print(chunk.text, end="", flush=True)

প্রতিটা অংশের কাজ
- for chunk in ...  : প্রতিটা টুকরো একটা একটা করে নেয়
- chunk.text        : টুকরোর ভেতরের লেখা
- end=""            : নতুন লাইন না দিয়ে একই লাইনে জুড়ে দেয় (ডিফল্টে end হলো নতুন লাইন)
- flush=True        : টুকরো আসামাত্র সাথে সাথে স্ক্রিনে দেখায়

flush=True কী করে
- কম্পিউটার সাধারণত ছাপার লেখা আগে একটা অস্থায়ী জায়গায় (buffer) জমায় এবং একসাথে পাঠায়
- flush=True সেই জমা রাখা বন্ধ করে, এখনই স্ক্রিনে পাঠাতে বাধ্য করে
- এটা না দিলে কিছু পরিবেশে লেখা জমে একসাথে হঠাৎ দেখা দিতে পারে, ফলে "একটু একটু করে ফোটা" অনুভূতি নষ্ট হয়

মনে রাখার বিষয়
- end="|" দিলে chunk-এর সীমানা দেখা যায় (শেখার জন্য ভালো)। স্বাভাবিক লেখার জন্য end="" দাও
- মডেল Markdown-এ (##, **, টেবিল) লিখলে print কাঁচা চিহ্নসহই দেখায়। সুন্দর রেন্ডারের জন্য display(Markdown(...)) ব্যবহার করা যায়
- সুবিধা: ইউজার অপেক্ষা না করে সাথে সাথে লেখা দেখে, তাই অ্যাপ দ্রুত মনে হয়

### Batch

batch is a collection of independent requests to a model that improve performace and reduce costs, as the processing can be done in parallel

```python
responses  = model.batch(["Why do parrots have colorful feathers",
             "How do airplace fly",
             "what is quantum computing?"])
for response in responses:
    print(response.content)
```

this can be written as below as well config:

```python
responses  = model.batch(["Why do parrots have colorful feathers",
             "How do airplace fly",
             "what is quantum computing?"],
            config ={
                "max_concurrency" : 5, # limit to 5 parallel calls
            }
                         )
for response in responses:
    print(response.content)
```

this is the end of streaming and batch

===== batch কেন গুরুত্বপূর্ণ =====

কী করে
- অনেকগুলো প্রশ্ন একসাথে (parallel) পাঠায়, উত্তরের তালিকা ফেরত দেয়
- উত্তর আসে প্রশ্নের একই ক্রমে (responses[0] = প্রথম প্রশ্নের উত্তর)

কেন দরকার
1. সময় বাঁচে
   - একটা একটা করে পাঠালে ১০০টা প্রশ্নে ১০০ গুণ সময় লাগে
   - batch-এ একসাথে অনেকগুলো চলে, তাই অনেক দ্রুত শেষ হয়

2. বাস্তব কাজে অনেক ইনপুট প্রসেস করতে হয়
   - শত শত রিভিউ ক্লাসিফাই করা (positive/negative)
   - অনেকগুলো ডকুমেন্টের সারাংশ বানানো
   - RAG বা agent-এর টেস্ট প্রশ্নের তালিকা একসাথে চালিয়ে evaluation করা

3. কোড সহজ থাকে
   - নিজে লুপ বা থ্রেড লিখতে হয় না, একটা লাইনেই কাজ হয়

নিয়ন্ত্রণ
- config={"max_concurrency": N} দিয়ে একসাথে সর্বোচ্চ কয়টা অনুরোধ চলবে তা ঠিক করা যায়

সতর্কতা
- প্রতিটা প্রশ্ন আলাদা API অনুরোধ, তাই ফ্রি টায়ারের সীমা (rate limit) দ্রুত শেষ হতে পারে
- 429 error এলে max_concurrency কমাও
- খরচ বাড়ে: যত প্রশ্ন, তত টোকেন ব্যবহার
- একটা অনুরোধ ব্যর্থ হলে পুরো batch-এ error আসতে পারে (return_exceptions=True দিলে ব্যর্থগুলো আলাদাভাবে ধরা যায়, ব্যবহারের আগে নিজের ভার্সনে যাচাই করে নিও)

এক লাইনে
- একটার বেশি ইনপুট থাকলে batch, একটা হলে invoke, লেখা একটু একটু করে দেখাতে হলে stream
