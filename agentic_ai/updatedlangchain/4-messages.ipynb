### Messages

Messages are the fundamental unit of context for models in Langchain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contains
* Role - identifies the message type
* Content - Represents the actual content of the message ( like text, audio , images, documents, etc.)
* Metadata - Optional fields such as response information  , messages IDs, and token usage

Langchain provides a standard message type that works across all model providers , ensuring consistent behaviour ragardless of the model being called

Whenever a model is giving an output its basically an AIMessage
Whenever a human is giving an input its basically a Human Message


```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] =  os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")
```

```python
model.invoke("Please tell me what is artificial intelligence?") #this is human message
# and the model will respond with a AIMessage, which is actually the output of the model.
```

## TEXT PROMPTS

Text prompts are strings, ideal for straightforward generation tasks,where you dont need to retain conversation history

```python
model.invoke("What is langchain?") #this is human message and we can also call this a text prompt.
```

Use text prompts when:
* You have a single , standalone request
* You dont need conversation history
* You want minimal code Complexity

### Message Prompts
Alternatively, you can pass in a list of messages to the model by providing a list of message objects.

Message types
* System message - Tells the model how to behave and provide context for interactions
* Human message - Represents user input and interactions with the model
* Ai message - Responses  generated by the model , including text content , tools calls and metadata
* Tool message - Represents the outputs of tool calls


#### System Message 
A systemMessage represents an initial set of instructions that primes the models behavior . You can use a system message to set the tone, define the model's role, and establish for responses.

#### Human Message
A human message represents use input and interactions. They can contain text, audio, files, and anh other amount of multimodal content

#### AI Message
AN AIMessage represents the output of a model invocation. They can include multimodal data, tool calls, and provider-specific metadata that you can later access

#### Tool Message
For models that support tool calling , AI messages can contain tool calls. Tool messages are used to pass the results of a single tool execution back to the model

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage


messages = [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on artificial intellegence")
    
]

response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are a helpful Coding Assistant")

messages = [
    system_msg,
    HumanMessage("How do i create a REST API")
]

response = model.invoke(messages)
print(response.content)
```

### Detailed info to the LLM through System Message

```python
from langchain.messages import SystemMessage, HumanMessage
system_msg = SystemMessage(
    """
    You are a senior Python developer with expertise in web frameworks.
    Always provide code examples and explain your reasoning.
    be precise but thorough in your explainations.
    """
)
messages = [
    system_msg,
    HumanMessage("How do I create a REST API?")
    
]

response = model.invoke(messages)
print(response.content)
```

### Message Metadata


```python
human_msg = HumanMessage(
content  = "Hello",
name = "alice", # optional : identify different users
id = "msg_123"  # Optional: unique identifier for tracing
)
```

```python
response  = model.invoke([human_msg])
response
```

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage
# create an AI message manually
ai_msg  = AIMessage("I'd be happy to help you with that question!")

# add to conversation history
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg , #insert as if it came from the model
    HumanMessage("Great! What's 2 + 2?")
]

response = model.invoke(messages)
print(response.content)
```

```python
# in order to see my metadata
response.usage_metadata
```

###  Tool message

```python
from langchain.messages import AIMessage, ToolMessage
# after a model makes a tool call 
# here we demonstrate manually creating the messages for brevity 

ai_message = AIMessage(
    content=[],
    tool_calls =[{
        "name": "get_weather",
        "args":{"location":"Mymensingh"},
        "id":"call_123"
    }]
)

# execute tool and create result message
weather_result = "Sunny, 72"
tool_messsage = ToolMessage(
    content = weather_result,
    tool_call_id = "call_123" #must match the call id
)


#continue conversation
messages = [
    HumanMessage("What is the weather in Mymensingh"),
    ai_message, # model's tool call
    tool_messsage, #Tool execution result
]

response =  model.invoke(messages)
response

```
